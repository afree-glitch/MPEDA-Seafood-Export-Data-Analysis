# MPEDA Seafood Export Data Analysis

Exploratory Data Analysis (EDA) of India's seafood export data from **2015-16 to 2024-2025**, focusing on export quantity, value, product-wise performance, trends, growth, correlations, and relationships using Python, Pandas, NumPy, Matplotlib, and Seaborn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
mpeda=pd.read_csv("/content/MPEDA_Item_Wise_Export_Data_2015-16_to_2024-25.csv")
mpeda
```

# Total Export Value

```python
plt.figure(figsize=(12, 6))

plt.plot(
    mpeda['Financial_Year'],
    mpeda['TOTAL_USD_MILLION'],
    marker='o',
    linewidth=2.5,
    color='royalblue'
)

plt.title('Total Seafood Export Value Over the Years',
          fontsize=16, fontweight='bold')

plt.xlabel('Financial Year')
plt.ylabel('Export Value (USD Million)')

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# Total Export Quantity

```python
plt.figure(figsize=(12, 6))

plt.plot(
    mpeda['Financial_Year'],
    mpeda['TOTAL_Q'],
    marker='o',
    linewidth=2.5,
    color='seagreen'
)

plt.title('Total Seafood Export Quantity Over the Years',
          fontsize=16, fontweight='bold')

plt.xlabel('Financial Year')
plt.ylabel('Export Quantity')

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# Average Export Quantity by Product

```python
quantity_cols = [
    'FROZEN SHRIMP_Q',
    'FROZEN FISH_Q',
    'FR CUTTLE FISH_Q',
    'FR SQUID_Q',
    'LIVE ITEMS_Q',
    'CHILLED ITEMS_Q',
    'OTHERS_Q'
]

quantity_mean = mpeda[quantity_cols].mean().sort_values()

plt.figure(figsize=(11, 6))

plt.barh(
    quantity_mean.index,
    quantity_mean.values,
    color='steelblue'
)

plt.title('Average Export Quantity by Product Category',
          fontsize=16, fontweight='bold')

plt.xlabel('Average Export Quantity')
plt.ylabel('Product Category')

plt.tight_layout()
plt.show()
```

# Average Export Value by Product

```python
value_cols = [
    'FROZEN SHRIMP_USD_MILLION',
    'FROZEN FISH_USD_MILLION',
    'FR CUTTLE FISH_USD_MILLION',
    'FR SQUID_USD_MILLION',
    'LIVE ITEMS_USD_MILLION',
    'CHILLED ITEMS_USD_MILLION',
    'OTHERS_USD_MILLION'
]

value_mean = mpeda[value_cols].mean().sort_values(ascending=False)

colors = [
    'seagreen',
    'orange',
    'tomato',
    'mediumpurple',
    'gold',
    'skyblue',
    'gray'
]

plt.figure(figsize=(12, 6))

plt.bar(
    value_mean.index,
    value_mean.values,
    color=colors
)

plt.title('Average Export Value by Product Category',
          fontsize=16, fontweight='bold')

plt.xlabel('Product Category')
plt.ylabel('Average Export Value (USD Million)')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

# Export Composition

```python
composition = mpeda[
    [
        'FROZEN SHRIMP_USD_MILLION',
        'FROZEN FISH_USD_MILLION',
        'FR CUTTLE FISH_USD_MILLION',
        'FR SQUID_USD_MILLION'
    ]
]

composition.index = mpeda['Financial_Year']

composition.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6),
    color=[
        'seagreen',
        'orange',
        'tomato',
        'mediumpurple'
    ]
)

plt.title('Composition of Seafood Export Value',
          fontsize=16, fontweight='bold')

plt.xlabel('Financial Year')
plt.ylabel('Export Value (USD Million)')

plt.xticks(rotation=45)
plt.legend(title='Product Category')

plt.tight_layout()
plt.show()
```

# Year-on-Year Growth

```python
mpeda['TOTAL_GROWTH_%'] = (
    mpeda['TOTAL_USD_MILLION'].pct_change() * 100
)

plt.figure(figsize=(12, 6))

plt.bar(
    mpeda['Financial_Year'],
    mpeda['TOTAL_GROWTH_%'],
    color='seagreen',edgecolor='black'
)

plt.axhline(
    0,
    color='black',
    linewidth=1
)

plt.title('Year-on-Year Growth in Seafood Export Value',
          fontsize=16, fontweight='bold')

plt.xlabel('Financial Year')
plt.ylabel('Growth (%)')

plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

# Frozen Shrimp Quantity vs Value

```python
plt.figure(figsize=(9, 6))

sns.regplot(
    data=mpeda,
    x='FROZEN SHRIMP_Q',
    y='FROZEN SHRIMP_USD_MILLION',
    scatter_kws={'color': 'seagreen', 's': 70},
    line_kws={'color': 'black', 'linewidth': 2}
)

plt.title('Frozen Shrimp Quantity vs Export Value',
          fontsize=16, fontweight='bold')

plt.xlabel('Frozen Shrimp Quantity')
plt.ylabel('Export Value (USD Million)')

plt.tight_layout()
plt.show()
```

# Correlation Heatmap

```python
corr = mpeda.select_dtypes(include='number').corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    corr,
    annot=True,
    fmt='.2f',
    cmap='RdYlBu_r',
    linewidths=0.5,
    center=0
)

plt.title('Correlation Matrix of Seafood Export Variables',
          fontsize=16, fontweight='bold')

plt.tight_layout()
plt.show()
```

# Distribution

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=mpeda,
    x='TOTAL_USD_MILLION',
    kde=True,
    color='royalblue',
    bins=8
)

plt.title('Distribution of Total Seafood Export Value',
          fontsize=16, fontweight='bold')

plt.xlabel('Total Export Value (USD Million)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

# Outlier Analysis

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=mpeda,
    y='TOTAL_USD_MILLION',
    color='skyblue'
)

plt.title('Distribution and Outliers of Total Export Value',
          fontsize=16, fontweight='bold')

plt.ylabel('Total Export Value (USD Million)')

plt.tight_layout()
plt.show()
```

# Frozen Shrimp Quantity & Value

```python
fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.plot(
    mpeda['Financial_Year'],
    mpeda['FROZEN SHRIMP_Q'],
    marker='o',
    linewidth=2.5,
    color='seagreen'
)

ax1.set_xlabel('Financial Year')
ax1.set_ylabel('Shrimp Quantity', color='seagreen')

ax1.tick_params(axis='y', labelcolor='seagreen')
ax1.tick_params(axis='x', rotation=45)

ax2 = ax1.twinx()

ax2.plot(
    mpeda['Financial_Year'],
    mpeda['FROZEN SHRIMP_USD_MILLION'],
    marker='s',
    linewidth=2.5,
    color='red'
)

ax2.set_ylabel(
    'Shrimp Export Value (USD Million)',
    color='red'
)

ax2.tick_params(axis='y', labelcolor='red')

plt.title(
    'Frozen Shrimp Export Quantity and Value',
    fontsize=16,
    fontweight='bold'
)

fig.tight_layout()
plt.show()
```

# Major Product Trends

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=mpeda,
    x='Financial_Year',
    y='FROZEN SHRIMP_USD_MILLION',
    marker='o',
    linewidth=2,
    color='seagreen',
    label='Frozen Shrimp'
)

sns.lineplot(
    data=mpeda,
    x='Financial_Year',
    y='FROZEN FISH_USD_MILLION',
    marker='o',
    linewidth=2,
    color='red',
    label='Frozen Fish'
)

sns.lineplot(
    data=mpeda,
    x='Financial_Year',
    y='FR CUTTLE FISH_USD_MILLION',
    marker='o',
    linewidth=2,
    color='black',
    label='Frozen Cuttlefish'
)

sns.lineplot(
    data=mpeda,
    x='Financial_Year',
    y='FR SQUID_USD_MILLION',
    marker='o',
    linewidth=2,
    color='purple',
    label='Frozen Squid'
)

plt.title(
    'Export Value Trends by Major Seafood Category',
    fontsize=16,
    fontweight='bold'
)

plt.xlabel('Financial Year')
plt.ylabel('Export Value (USD Million)')

plt.xticks(rotation=45)
plt.legend()
plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```
